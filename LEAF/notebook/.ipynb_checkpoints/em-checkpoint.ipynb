# Energy management example

This notebook walks through the energy-management example in three steps: define the problem formulation, train a learned Moreau envelope (LME) model, and benchmark inference inside ADMM.

The goal is to make the workflow readable for non-experts. Most heavy computations are wrapped behind flags, so the notebook can first be read and inspected without launching long optimization or training jobs.

## Code map

- `setup_em.jl` defines the MPC data, forecasts, constraints, and economic cost.
- `generate_em.jl` solves ADMM subproblems and saves training targets in `.npz` files.
- `lme/ICNN_em.py` defines the input-convex neural network (ICNN) and its training loss.
- `lme/train_em.py` loads datasets, trains the ICNN, and exports a JSON model for Julia.
- `benchmarks_em.jl` compares standard ADMM with learned-envelope ADMM variants.

```julia
from pathlib import Path
import json
import subprocess
import sys

import numpy as np

# The notebook lives in src/LEAF/examples/energy_management/notebook.
NOTEBOOK_DIR = Path.cwd()
if NOTEBOOK_DIR.name == 'notebook':
    EM_DIR = NOTEBOOK_DIR.parent
else:
    EM_DIR = Path('src/LEAF/examples/energy_management').resolve()

REPO_ROOT = EM_DIR.parents[3]
DATA_DIR = EM_DIR / 'datasets'
MODEL_DIR = EM_DIR / 'model'
LME_DIR = EM_DIR / 'lme'

print(f'Energy-management directory: {EM_DIR}')
print(f'Dataset directory: {DATA_DIR}')
print(f'Model directory: {MODEL_DIR}')
```

# Problem setup

We control a battery over a 24-hour horizon with 15-minute time steps. At each time step, the controller decides how much battery power to charge or discharge and how much demand to serve. The forecasts come from solar generation and building load data.

## MPC formulation in plain language

For each time step `k`, the model uses these main variables:

- `m[k]`: net grid exchange.
- `u[k]`: battery power action.
- `p[k]`: served power entering the discomfort penalty.
- `x[k]`: battery state of charge.

The core constraints are:

$$x_{k+1} = x_k - \Delta t\,u_k / BESS$$

$$u_k + m_k + generation_k - load_k - p_k = 0$$

The first equation updates the battery state. The second equation balances power: battery action, grid exchange, solar generation, load, and served power must agree.

The stage cost combines three practical goals:

- Energy cost: discourage expensive grid exchange.
- Battery operating cost: discourage unnecessary battery movement.
- Discomfort cost: discourage serving too little power.

```julia
load_csv = np.genfromtxt(DATA_DIR / 'load_15min_max100kW_SanDiego_Building.csv', delimiter=',', skip_header=1)
gen_csv = np.genfromtxt(DATA_DIR / 'PV_48h_15-min_150kW_San_Diego.csv', delimiter=',', skip_header=1)

load = load_csv[:96, 1]
gen = gen_csv[:96, 1]
time_hours = np.arange(96) * 0.25

try:
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=(9, 3.5))
    ax.plot(time_hours, load, label='Building load')
    ax.plot(time_hours, gen, label='Solar generation')
    ax.set_xlabel('Time (hours)')
    ax.set_ylabel('Power (kW)')
    ax.set_title('One-day forecasts used by the MPC problem')
    ax.grid(True, alpha=0.3)
    ax.legend()
    plt.show()
except ImportError:
    preview = np.column_stack([time_hours[:8], load[:8], gen[:8]])
    print('Install matplotlib to see the line plot. Forecast preview:')
    print('hour    load_kW    solar_kW')
    for hour, load_kw, solar_kw in preview:
        print(f'{hour:4.2f}  {load_kw:8.3f}  {solar_kw:8.3f}')
```

## Generating training data

`generate_em.jl` creates training examples by running ADMM for several initial battery states. During ADMM, it records three arrays:

- `input`: the point where the envelope is evaluated.
- `enve`: the Moreau envelope value.
- `grad`: the gradient of the envelope with respect to the input.

These are exactly the supervised targets used to train the ICNN. The full generation run can take time because it repeatedly solves optimization subproblems.

```julia
# Set this to True when you want to regenerate the .npz files from scratch.
RUN_DATA_GENERATION = False

if RUN_DATA_GENERATION:
    subprocess.run(
        ['julia', '--project=.', str(EM_DIR / 'generate_em.jl')],
        cwd=REPO_ROOT,
        check=True,
    )
else:
    print('Skipping data generation. Using the checked-in .npz files in datasets/.')
```

```julia
train_path = DATA_DIR / 'eco_mpc-rho=1.0-train.npz'
test_path = DATA_DIR / 'eco_mpc-rho=1.0-test.npz'

with np.load(train_path) as data:
    X_train = data['input'].T
    y_train = data['enve']
    g_train = data['grad'].T
    rho = data['rho'].item()

with np.load(test_path) as data:
    X_test = data['input'].T
    y_test = data['enve']
    g_test = data['grad'].T

print(f'rho: {rho}')
print(f'training inputs: {X_train.shape}')
print(f'training values: {y_train.shape}')
print(f'training gradients: {g_train.shape}')
print(f'test inputs: {X_test.shape}')
print(f'input dimension per time step: {X_train.shape[1]}')
```

# Train model

The learned model is an input-convex neural network (ICNN). Convexity matters here because ADMM uses the model as a surrogate for a convex optimization object. The ICNN is trained with two signals at the same time:

- match the envelope value `enve`, and
- match the envelope gradient `grad`.

The training script `lme/train_em.py` uses the reusable functions in `lme/ICNN_em.py`, then exports the trained parameters to `model/neco_mpc-rho=1.json` for Julia inference.

```julia
sys.path.insert(0, str(LME_DIR))

from train_em import evaluate, load_dataset, prepare_for_export, save_model
from ICNN_em import train_icnn

# Keep the notebook quick by default. Set True to reproduce the full training script.
RUN_FULL_TRAINING = False

if RUN_FULL_TRAINING:
    subprocess.run([sys.executable, str(LME_DIR / 'train_em.py')], cwd=REPO_ROOT, check=True)
else:
    Xtr, ytr, gtr, rho = load_dataset(train_path)
    Xva, yva, gva, _ = load_dataset(test_path)

    # A small smoke-training run demonstrates the exact API without waiting for 5000 epochs.
    demo_n = min(64, Xtr.shape[0])
    params = train_icnn(
        Xtr[:demo_n],
        ytr[:demo_n],
        gtr[:demo_n],
        n_in=Xtr.shape[1],
        widths=[16, 16],
        lr=1e-3,
        grad_weight=5.0,
        batch_size=16,
        epochs=20,
        seed=0,
    )

    val_mse, grad_mse = evaluate(params, Xva[:demo_n], yva[:demo_n], gva[:demo_n])
    print(f'Demo value MSE: {float(val_mse):.4e}')
    print(f'Demo grad MSE: {float(grad_mse):.4e}')
    print('For the paper-style model, set RUN_FULL_TRAINING=True or run lme/train_em.py directly.')
```

```julia
model_path = MODEL_DIR / 'neco_mpc-rho=1.json'
with open(model_path) as f:
    exported_model = json.load(f)

print(f'Exported model: {model_path.name}')
print(f'rho: {exported_model.get("rho")}')
print(f'number of ICNN layers: {len(exported_model["U"])}')
print('input-weight shapes:', [np.asarray(w).shape for w in exported_model['U']])
print('state-weight shapes:', [np.asarray(w).shape for w in exported_model['W']])
```

# Inference

At inference time, Julia loads the exported ICNN and uses its gradient inside the ADMM update. The benchmark compares:

- standard ADMM, which solves the proximal update with optimization solvers;
- LME-ADMM, which replaces that update with the learned envelope gradient;
- split LME-ADMM, which separates dynamics and bound projections for a faster structured update.

The benchmark script writes timing arrays and figures under the energy-management example directory.

```julia
# Set this to True to run the full Julia benchmark.
RUN_BENCHMARK = False

if RUN_BENCHMARK:
    subprocess.run(
        ['julia', '--project=.', str(EM_DIR / 'benchmarks_em.jl')],
        cwd=REPO_ROOT,
        check=True,
    )
else:
    print('Skipping full benchmark. Set RUN_BENCHMARK=True to run benchmarks_em.jl.')
```

```julia
julia_code = textwrap.dedent(r'''
    include("src/LEAF/examples/energy_management/setup_em.jl")
    include("src/LEAF/examples/energy_management/preprocess.jl")
    include("src/LEAF/examples/energy_management/solver_em.jl")
    include("src/LEAF/examples/energy_management/admm_em.jl")

    data = energy_mag()
    load = Vector{Float64}(data.load_forecast[1:data.N])
    gen = Vector{Float64}(data.gen_forecast[1:data.N])

    prime_sol = prime_sol_struct("Ipopt", data)
    aux_sol = aux_solver_eco("Clarabel", data)
    admm_sol = ADMM_eco_iter(data, prime_sol, aux_sol; max_iter=20)

    sol, solve_time, objective = admm_sol(data.x0, load, gen, ADMM_callback; verbose=false)
    println("Quick ADMM check with 20 max iterations")
    println("Objective: ", objective)
    println("Measured solver time: ", 1000 * solve_time, " ms")
    println("Solution shape: ", size(sol))
''')

RUN_QUICK_ADMM_CHECK = False

if RUN_QUICK_ADMM_CHECK:
    result = subprocess.run(
        ['julia', '--project=.', '-e', julia_code],
        cwd=REPO_ROOT,
        text=True,
        capture_output=True,
    )
    print(result.stdout)
    if result.returncode != 0:
        print(result.stderr)
        raise RuntimeError('Quick ADMM check failed')
else:
    print('Quick ADMM check is disabled by default because it builds optimization models.')
```

## How to read the benchmark

The important outputs are the objective gap and the timing distribution. A useful learned model should keep the objective close to the high-accuracy baseline while reducing repeated solve time. The timing plot produced by `benchmarks_em.jl` summarizes this tradeoff across repeated runs.
